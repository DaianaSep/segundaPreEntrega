# BLOQUE 4: LÓGICA ALGORÍTMICA Y FUNCIONES PERSONALIZADAS (PYTHON PURO)
# ==========================================
Implementación y aplicación de funciones personalizadas (`def` y `lambda`) sobre el DataFrame.


```python
# 4.1. FUNCIÓN FORMAL (def)
# Definición de función con estructura condicional (if/elif/else) basada en la regla de negocio
# para categorizar el nivel de jornada laboral según las horas de trabajo diario.

def categorizar_jornada(horas):
    # Comentario explicativo: Evaluamos el rango de horas de trabajo para clasificar la sobrecarga laboral
    if pd.isna(horas):
        return 'Sin Datos'
    elif horas > 10:
        return 'Jornada Excesiva'
    elif horas >= 8:
        return 'Jornada Completa'
    else:
        return 'Jornada Reducida'

# Aplicación de la función formal usando .apply() para generar una nueva columna
df['categoria_jornada'] = df['daily_work_hours'].apply(categorizar_jornada)

# Verificación del resultado
display(df['categoria_jornada'].value_counts(dropna=False))
```

```python
# 4.2. FUNCIÓN EXPRESS (lambda)
# Función lambda para realizar una transformación rápida de texto: normalizar la columna 'burnout_level' a minúsculas.

df['burnout_level_minusc'] = df['burnout_level'].apply(lambda x: x.lower() if isinstance(x, str) else x)

# Muestra del resultado transformado
display(df[['burnout_level', 'burnout_level_minusc', 'categoria_jornada']].head())
```
